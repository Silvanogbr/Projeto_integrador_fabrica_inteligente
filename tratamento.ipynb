```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Bibliotecas importadas com sucesso!")
```

```python

# Carregar os dados dos sensores industriais
df = pd.read_csv("dados_sensores_metaltech.csv")

print("Arquivo carregado com sucesso!")
print(f"Quantidade de linhas: {df.shape[0]}")
print(f"Quantidade de colunas: {df.shape[1]}")
```

```python

# Converte a coluna de tempo para o formato datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Ordenar os registros por máquina e tempo
df = df.sort_values(["id_maquina", "timestamp"]).reset_index(drop=True)

print("Dados ordenados por máquina e tempo.")
display(df.head())
```

```python

# Colunas numéricas dos sensores
sensores = [
    "temperatura_c",
    "vibracao_mm_s",
    "pressao_bar"
]

# Interpolar os valores ausentes separadamente por máquina
for coluna in sensores:
    df[coluna] = df.groupby("id_maquina")[coluna].transform(
        lambda serie: serie.interpolate(limit_direction="both")
    )

print("Valores ausentes após a interpolação:")
print(df[sensores].isnull().sum())
```

```python

print("Total de valores ausentes no dataset:")
print(df.isnull().sum())

print("\nTotal de valores ausentes:")
print(df.isnull().sum().sum())
```

```python

# Verificar estatísticas das variáveis dos sensores
display(df[sensores].describe())

# Contar temperaturas suspeitas
print("Temperaturas iguais a 999.9:")
print((df["temperatura_c"] == 999.9).sum())

# Contar vibrações negativas
print("\nVibrações negativas:")
print((df["vibracao_mm_s"] < 0).sum())

# Contar pressões negativas
print("\nPressões negativas:")
print((df["pressao_bar"] < 0).sum())
```

```python

# Recarregar os dados originais para refazer o tratamento corretamente
df = pd.read_csv("dados_sensores_metaltech.csv")

# Converter a coluna de tempo
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Ordenar por máquina e tempo
df = df.sort_values(
    ["id_maquina", "timestamp"]
).reset_index(drop=True)

# Definir as colunas dos sensores
sensores = [
    "temperatura_c",
    "vibracao_mm_s",
    "pressao_bar"
]

# Transformar medições inválidas em valores ausentes
df.loc[df["temperatura_c"] == 999.9, "temperatura_c"] = np.nan
df.loc[df["vibracao_mm_s"] < 0, "vibracao_mm_s"] = np.nan

# Interpolar os valores ausentes por máquina
for coluna in sensores:
    df[coluna] = df.groupby("id_maquina")[coluna].transform(
        lambda serie: serie.interpolate(limit_direction="both")
    )

# Conferir os resultados
print("Valores ausentes após o tratamento:")
print(df.isnull().sum())

print("\nTemperaturas iguais a 999.9:",
      (df["temperatura_c"] == 999.9).sum())

print("Vibrações negativas:",
      (df["vibracao_mm_s"] < 0).sum())

print("\nDimensões finais:", df.shape)
```

```python

# Criar boxplots das variáveis dos sensores
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, coluna in zip(axes, sensores):
    sns.boxplot(y=df[coluna], ax=ax)
    ax.set_title(coluna)

plt.suptitle("Boxplots das variáveis dos sensores")
plt.tight_layout()
plt.show()
```

```python

# Histogramas das variáveis dos sensores
df[sensores].hist(
    figsize=(12, 8),
    bins=30,
    edgecolor="black"
)

plt.suptitle("Distribuição das variáveis dos sensores")
plt.tight_layout()
plt.show()
```

```python

# Selecionar as variáveis para a análise de correlação
colunas_correlacao = [
    "temperatura_c",
    "vibracao_mm_s",
    "pressao_bar",
    "falha_motor"
]

# Calcular a matriz de correlação
matriz_correlacao = df[colunas_correlacao].corr()

# Gerar o mapa de calor
plt.figure(figsize=(8, 6))

sns.heatmap(
    matriz_correlacao,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Matriz de Correlação dos Sensores e Falhas")
plt.tight_layout()
plt.show()
```
